```python
import duckdb
from pathlib import Path
import math

DB_PATH = r"J:\cms\Internal\Territories\EMEA\01 Regions\Italy-Iberia\08 Tools\Data Augmentation - Height\GRe Database by Region\CARTO\GRe_AUG_Database_CARTO_Piemonte_lookup.duckdb"
TEST_POINTS = [
    (9.1533, 45.18568586),   # lon, lat
]

TABLE_NAME = "buildings"
LON_COL = "centroid_lon"
LAT_COL = "centroid_lat"

SEARCH_RADIUS_M = 5000
MAX_RESULTS = 10


def q(name):
    return '"' + str(name).replace('"', '""') + '"'


con = duckdb.connect(DB_PATH, read_only=True)

try:
    con.execute("LOAD spatial;")
except Exception:
    pass

# Get available columns
available_cols = con.execute(f"""
    SELECT column_name
    FROM information_schema.columns
    WHERE table_name = '{TABLE_NAME}'
    ORDER BY ordinal_position
""").fetchdf()["column_name"].tolist()

preferred_show_cols = [
    "building_id",
    "source",
    "relation_id",
    "REGION",
    "CITY",
    "PROVINCE",
    "state",
    "postcode",
    "quadkey",
    "quadkey_prefix_6",
    "quadkey_prefix_14",
    LON_COL,
    LAT_COL,
    "footprint_area_m2",
    "height_m",
    "stories_exact",
    "occupancy_raw",
    "occupancy_group",
]

show_cols = []
for col in preferred_show_cols:
    if col in available_cols and col not in show_cols:
        show_cols.append(col)

print("Showing columns:")
print(show_cols)

show_cols_sql = ",\n                ".join(q(c) for c in show_cols)

for lon, lat in TEST_POINTS:
    lat_delta = SEARCH_RADIUS_M / 111_320.0
    lon_delta = SEARCH_RADIUS_M / (111_320.0 * max(math.cos(math.radians(lat)), 0.2))

    print("\n" + "=" * 100)
    print(f"Input coordinate: lon={lon}, lat={lat}")
    print(f"Search radius: {SEARCH_RADIUS_M:,} m")
    print("=" * 100)

    query = f"""
        WITH candidates AS (
            SELECT
                {show_cols_sql},
                TRY_CAST({q(LON_COL)} AS DOUBLE) AS __lon,
                TRY_CAST({q(LAT_COL)} AS DOUBLE) AS __lat
            FROM {q(TABLE_NAME)}
            WHERE
                TRY_CAST({q(LON_COL)} AS DOUBLE) BETWEEN {lon - lon_delta} AND {lon + lon_delta}
                AND TRY_CAST({q(LAT_COL)} AS DOUBLE) BETWEEN {lat - lat_delta} AND {lat + lat_delta}
                AND TRY_CAST({q(LON_COL)} AS DOUBLE) BETWEEN -180 AND 180
                AND TRY_CAST({q(LAT_COL)} AS DOUBLE) BETWEEN -90 AND 90
        )
        SELECT
            *,
            6371000 * 2 * ASIN(
                SQRT(
                    POWER(SIN(RADIANS((__lat - {lat}) / 2)), 2)
                    +
                    COS(RADIANS({lat}))
                    * COS(RADIANS(__lat))
                    * POWER(SIN(RADIANS((__lon - {lon}) / 2)), 2)
                )
            ) AS distance_m
        FROM candidates
        ORDER BY distance_m
        LIMIT {MAX_RESULTS}
    """

    df = con.execute(query).fetchdf()
    display(df)

    if df.empty:
        print("No buildings found within search radius.")
    else:
        print(f"Nearest distance: {df.iloc[0]['distance_m']:,.2f} m")

con.close()
```
